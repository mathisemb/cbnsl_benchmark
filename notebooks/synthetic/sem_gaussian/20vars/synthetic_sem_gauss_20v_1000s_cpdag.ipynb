# Synthetic medium SEM dataset (Gaussian)

- Gaussian linear SEM
- 20 variables
- 1000 samples

(comparison with the cpdags)

```python
import sys; sys.path.insert(0, "../../../..")
import pyagrum as gum
from pipeline.Benchmark import Benchmark
 
dag = gum.DAG()
dag.addNodes(20)
for t, h in [(1,0),(0,2),(2,3),(1,3),(3,4),(4,5),(4,6),(5,7),(6,7),(6,8),(9,8),(2,9),(11,7),(0,15),
             (9,10),(12,10),(12,11),(16,15),(14,15),(16,14),(13,12),(14,13),(13,17),(16,17),(17,18),
             (18,19)]:
    dag.addArc(t, h)

bench = Benchmark.synthetic_gausslinSEM(dag, n_samples=1000, seed=42)
```

## 1. Golden structure

```python
bench.show_golden()
```

## 2. Grid search

```python
bench.run()
```

```python
bench.plot_grid_search()
```
