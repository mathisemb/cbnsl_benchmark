# Synthetic basic CBN dataset

- uniform marginals
- normal local conditional copulas
- 5 variables
- 2000 samples

(comparison with the skeletons)

```python
import sys; sys.path.insert(0, "../../../..")
import pyagrum as gum
from pipeline.Benchmark import Benchmark
 
dag = gum.DAG()
dag.addNodes(5)
for t, h in [(1,0),(1,2),(3,2),(3,4),(4,2)]:
    dag.addArc(t, h)

bench = Benchmark.synthetic_cbn_unif_gauss(dag, n_samples=2000, seed=42, compare_mode="skeleton")
```

```python
bench.show_golden()
```
