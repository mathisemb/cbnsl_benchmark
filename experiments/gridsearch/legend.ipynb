```python
%config InlineBackend.figure_formats = ['svg']

import matplotlib.pyplot as plt
import matplotlib as mpl

# True LaTeX rendering
mpl.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "font.serif": ["Computer Modern"],
    "text.latex.preamble": r"\usepackage{lmodern}",
})

fig, ax = plt.subplots(figsize=(5, 1.8))
ax.set_xlim(0, 10)
ax.set_ylim(0.5, 4.5)
ax.axis("off")

entries = [
    (4.0, "forestgreen", "-",  2.5, "Correct arc/edge"),
    (3.2, "red",         "-",  2.5, "False positive arc/edge"),
    (2.4, "orange",      "-",  2.5, "Reversed arc"),
    (1.6, "grey",        "--", 1.5, "Missing arc/edge (false negative)"),
]

for y, color, style, lw, label in entries:
    ax.plot([1.0, 3.0], [y, y], color=color, linestyle=style, linewidth=lw, solid_capstyle="round")
    ax.text(3.4, y, label, va="center", fontsize=11)

plt.tight_layout()
plt.savefig("legend.pdf", bbox_inches="tight", pad_inches=0.1)
plt.show()
```
