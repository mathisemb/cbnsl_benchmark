# CMIIC : seuil historique contre seuil analytique

Même grille, mêmes graphes et mêmes jeux de données que le scaling de juillet 2026 (seed 42, 30 graphes).
`CPC` et `CMIIC` sont le run de juillet (mode Naive, pénalité α = 0.05). `CMIIC seuil analytique` est le run du 29 octobre 2026 (mode Analytic, split, seuil μ_n + z σ_tot, α = 0.05).

Mêmes figures que la section scaling du rapport : temps d'exécution, puis SHD, F1 et TPR sur les CPDAG, puis sur les squelettes.

```python
import sys; sys.path.insert(0, "../../..")
import pandas as pd

from experiments.scaling.io import load_results
from experiments.scaling.plot import plot_facet

JULY = {
    "sem_gaussian": "2026-07-30_14h34_sem_gaussian",
    "sem_laplace": "2026-07-30_23h47_sem_laplace",
    "cbn_unif_gauss": "2026-07-31_02h24_cbn_unif_gauss",
    "cbn_exp_clayton": "2026-07-31_06h33_cbn_exp_clayton",
    "cbn_unif_mixture": "2026-07-31_10h35_cbn_unif_mixture"
}
NEW = {
    "sem_gaussian": "2026-10-29_18h55_sem_gaussian",
    "sem_laplace": "2026-10-29_20h19_sem_laplace",
    "cbn_unif_gauss": "2026-10-29_21h38_cbn_unif_gauss",
    "cbn_exp_clayton": "2026-10-30_00h46_cbn_exp_clayton",
    "cbn_unif_mixture": "2026-10-30_03h53_cbn_unif_mixture"
}

LABELS = {"CPC": "CPC", "CMIIC": "CMIIC", "CMIIC_analytic": "CMIIC seuil analytique"}

def load_pair(gen):
    """Lignes CPC et CMIIC du run de juillet et run du seuil analytique, pour un générateur."""
    old = load_results(f"{JULY[gen]}/results.csv")
    old = old[old.algo.isin(["CPC", "CMIIC"])]
    new = load_results(f"{NEW[gen]}/results.csv")
    df = pd.concat([old, new], ignore_index=True)
    df["algo"] = df.algo.map(LABELS)
    return df

CPDAG = ["SHD", "F1-Score", "TPR"]
SKELETON = [f"{m}_skeleton" for m in CPDAG]
```
